# Waste Volume Forecasting - Training

Trains an LSTM model to forecast future waste volume based on historical time-series data, used for capacity planning of the carbon-conversion process.

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from data_loader import load_series, create_windows
from model import build_model
```

## Load and prepare data

```python
WINDOW_SIZE = 30
HORIZON = 7

df = load_series("data/waste_volume.csv")
scaler = MinMaxScaler()
scaled = scaler.fit_transform(df[["waste_volume"]]).flatten()

X, y = create_windows(scaled, window_size=WINDOW_SIZE, horizon=HORIZON)
X = X.reshape((X.shape[0], X.shape[1], 1))

split = int(len(X) * 0.8)
X_train, X_val = X[:split], X[split:]
y_train, y_val = y[:split], y[split:]
print(f"Train: {X_train.shape}, Val: {X_val.shape}")
```

## Build and train model

```python
model = build_model(window_size=WINDOW_SIZE, horizon=HORIZON)
model.summary()
```

```python
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=16,
)
```

## Save model

```python
model.save("models/waste_forecast_lstm.h5")
print("Model saved to models/waste_forecast_lstm.h5")
```
