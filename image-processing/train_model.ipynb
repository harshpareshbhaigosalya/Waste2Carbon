# Waste Image Classification - Training

Trains a CNN to classify waste images into material categories (organic, plastic, metal, paper, glass) used upstream of the carbon-conversion estimation pipeline.

```python
import numpy as np
from sklearn.model_selection import train_test_split
from preprocess import load_dataset
from model import build_model
```

## Load dataset

```python
DATA_DIR = "data"
images, labels = load_dataset(DATA_DIR)

class_names = sorted(set(labels))
label_to_index = {name: i for i, name in enumerate(class_names)}
y = np.array([label_to_index[l] for l in labels])

X_train, X_val, y_train, y_val = train_test_split(
    images, y, test_size=0.2, random_state=42, stratify=y
)
print(f"Train: {X_train.shape}, Val: {X_val.shape}")
```

## Build and train model

```python
model = build_model(num_classes=len(class_names))
model.summary()
```

```python
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=20,
    batch_size=32,
)
```

## Save model

```python
model.save("models/waste_image_classifier.h5")
print("Model saved to models/waste_image_classifier.h5")
```
